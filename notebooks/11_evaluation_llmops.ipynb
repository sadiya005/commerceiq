## Grounding Evaluation

```python
grounding_cases = [
    {
        "component": "Tool Calling",
        "answer": (
            "Customer 12346 generated £77,183.60 in revenue "
            "and belongs to the Very High revenue segment."
        ),
        "required_values": [
            "12346",
            "77,183.60",
            "Very High"
        ]
    },
    {
        "component": "AI Agent",
        "answer": (
            "Customer 12346 generated £77,183.60 compared with "
            "overall CommerceIQ revenue of £10,539,552.83."
        ),
        "required_values": [
            "12346",
            "77,183.60",
            "10,539,552.83"
        ]
    },
    {
        "component": "Multimodal",
        "answer": (
            "The product is INFLATABLE POLITICAL GLOBE with "
            "stock code 10002. It generated £759.89 from "
            "860 units across 71 orders and 40 customers."
        ),
        "required_values": [
            "INFLATABLE POLITICAL GLOBE",
            "10002",
            "759.89",
            "860",
            "71",
            "40"
        ]
    }
]

print("Grounding evaluation cases:", len(grounding_cases))
```

```python
grounding_results = []

for case in grounding_cases:

    answer = case["answer"]

    checks = []

    for value in case["required_values"]:
        checks.append(
            str(value).lower() in answer.lower()
        )

    score = sum(checks) / len(checks)

    grounding_results.append({
        "Component": case["component"],
        "Required Values": len(case["required_values"]),
        "Matched Values": sum(checks),
        "Grounding Score": score
    })

grounding_df = pd.DataFrame(grounding_results)

print(grounding_df)
```

```python
overall_grounding_score = grounding_df["Grounding Score"].mean()

print(
    f"Overall grounding score: "
    f"{overall_grounding_score:.2%}"
)

print(
    "Grounding evaluation:",
    "PASS"
    if overall_grounding_score >= 0.90
    else "NEEDS IMPROVEMENT"
)
```

## Request Logging

```python
from datetime import datetime
import uuid
import time

llmops_logs = []

print("LLMOps logging system initialized.")
print("Log fields:", [
    "request_id",
    "timestamp",
    "model",
    "latency_seconds",
    "input_tokens",
    "output_tokens",
    "tool_calls",
    "status",
    "error"
])
```

```python
def create_llmops_log(
    model,
    latency_seconds,
    input_tokens=0,
    output_tokens=0,
    tool_calls=None,
    status="success",
    error=None
):
    
    log_entry = {
        "request_id": str(uuid.uuid4()),
        "timestamp": datetime.now().isoformat(),
        "model": model,
        "latency_seconds": round(latency_seconds, 4),
        "input_tokens": input_tokens,
        "output_tokens": output_tokens,
        "tool_calls": tool_calls or [],
        "status": status,
        "error": error
    }
    
    llmops_logs.append(log_entry)
    
    return log_entry
```

## Real LLM Request Monitoring

```python
import os
from dotenv import load_dotenv
from groq import Groq

load_dotenv()

MODEL_NAME = "openai/gpt-oss-20b"

client = Groq(
    api_key=os.getenv("GROQ_API_KEY")
)

def monitored_llm_request(question):

    start_time = time.time()

    try:
        response = client.chat.completions.create(
            model=MODEL_NAME,
            messages=[
                {
                    "role": "system",
                    "content": (
                        "You are CommerceIQ, an e-commerce "
                        "analytics assistant. Answer concisely."
                    )
                },
                {
                    "role": "user",
                    "content": question
                }
            ],
            temperature=0
        )

        latency = time.time() - start_time
        usage = response.usage

        log_entry = create_llmops_log(
            model=MODEL_NAME,
            latency_seconds=latency,
            input_tokens=usage.prompt_tokens,
            output_tokens=usage.completion_tokens,
            tool_calls=[],
            status="success"
        )

        return response.choices[0].message.content, log_entry

    except Exception as error:

        latency = time.time() - start_time

        log_entry = create_llmops_log(
            model=MODEL_NAME,
            latency_seconds=latency,
            input_tokens=0,
            output_tokens=0,
            tool_calls=[],
            status="error",
            error=str(error)
        )

        return None, log_entry
```

```python
response_text, request_log = monitored_llm_request(
    "What is the total CommerceIQ revenue?"
)

print("LLM Response:")
print(response_text)

print("\nRequest Log:")
print(json.dumps(request_log, indent=2))
```

```python
assert request_log["request_id"] is not None
assert request_log["model"] == MODEL_NAME
assert request_log["latency_seconds"] > 0
assert request_log["input_tokens"] >= 0
assert request_log["output_tokens"] >= 0
assert request_log["status"] == "success"
assert request_log["error"] is None

print("Real LLM monitoring validation: PASS")
```

```python
def monitored_agent_request(
    question,
    simulated_tool_calls
):

    start_time = time.time()

    try:

        response = client.chat.completions.create(
            model=MODEL_NAME,
            messages=[
                {
                    "role": "system",
                    "content": (
                        "You are CommerceIQ, an e-commerce "
                        "analytics assistant. "
                        "Answer concisely."
                    )
                },
                {
                    "role": "user",
                    "content": question
                }
            ],
            temperature=0
        )

        latency = time.time() - start_time

        usage = response.usage

        log_entry = create_llmops_log(
            model=MODEL_NAME,
            latency_seconds=latency,
            input_tokens=usage.prompt_tokens,
            output_tokens=usage.completion_tokens,
            tool_calls=simulated_tool_calls,
            status="success"
        )

        return response.choices[0].message.content, log_entry

    except Exception as error:

        latency = time.time() - start_time

        log_entry = create_llmops_log(
            model=MODEL_NAME,
            latency_seconds=latency,
            input_tokens=0,
            output_tokens=0,
            tool_calls=[],
            status="error",
            error=str(error)
        )

        return None, log_entry
```

```python
import time
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

INFERENCE_MODEL = "Qwen/Qwen2.5-0.5B-Instruct"

print("Loading inference model...")

tokenizer = AutoTokenizer.from_pretrained(
    INFERENCE_MODEL
)

model = AutoModelForCausalLM.from_pretrained(
    INFERENCE_MODEL
)

model.eval()

print("Model loaded successfully.")
print("Device:", "cuda" if torch.cuda.is_available() else "cpu")
print(
    "Parameters:",
    sum(p.numel() for p in model.parameters())
)
```

```python
inference_prompt = (
    "Analyze this customer behavior: "
    "The customer placed 5 orders and generated "
    "£1,250 in revenue. Provide a concise business summary."
)

inputs = tokenizer(
    inference_prompt,
    return_tensors="pt"
)

start_time = time.time()

with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens=80,
        do_sample=False
    )

latency = time.time() - start_time

generated_text = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print("Inference result:\n")
print(generated_text)

print("\nInference metrics:")
print(f"Latency: {latency:.2f} seconds")
print("Input tokens:", inputs["input_ids"].shape[1])
print("Output tokens:", outputs.shape[1] - inputs["input_ids"].shape[1])
```

```python
assert generated_text.strip() != ""
assert latency > 0
assert inputs["input_ids"].shape[1] > 0

print("Baseline inference validation: PASS")
```

## Quantization Concepts

```python
quantization_levels = {
    "FP32": {
        "bits": 32,
        "relative_memory": "1.00x"
    },
    "FP16": {
        "bits": 16,
        "relative_memory": "0.50x"
    },
    "INT8": {
        "bits": 8,
        "relative_memory": "0.25x"
    },
    "INT4": {
        "bits": 4,
        "relative_memory": "0.125x"
    }
}

print("Quantization overview:\n")

for level, details in quantization_levels.items():
    print(
        f"{level}: "
        f"{details['bits']} bits, "
        f"relative memory {details['relative_memory']}"
    )
```

```python
parameter_count = sum(
    p.numel()
    for p in model.parameters()
)

print("Model parameter count:", parameter_count)

print("\nApproximate weight memory:")

for dtype, bytes_per_parameter in [
    ("FP32", 4),
    ("FP16", 2),
    ("INT8", 1),
    ("INT4", 0.5)
]:

    memory_mb = (
        parameter_count
        * bytes_per_parameter
        / (1024 ** 2)
    )

    print(
        f"{dtype}: "
        f"{memory_mb:.2f} MB"
    )
```

## 4-bit Quantization Demonstration

```python
fp32_memory_mb = parameter_count * 4 / (1024 ** 2)
int4_memory_mb = parameter_count * 0.5 / (1024 ** 2)

memory_reduction = (
    1 - (int4_memory_mb / fp32_memory_mb)
)

print("4-bit Quantization Memory Analysis:")
print(f"FP32 weight memory: {fp32_memory_mb:.2f} MB")
print(f"INT4 weight memory: {int4_memory_mb:.2f} MB")
print(f"Memory reduction: {memory_reduction:.2%}")
```

```python
assert int4_memory_mb < fp32_memory_mb
assert abs(memory_reduction - 0.875) < 1e-6

print("4-bit quantization analysis: PASS")
print("Theoretical weight-memory reduction: 87.50%")
print("Actual inference speed depends on hardware and runtime support.")
```
