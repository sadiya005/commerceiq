```python
df_clean["Invoice"] = df_clean["Invoice"].astype(str)
df_clean["StockCode"] = df_clean["StockCode"].astype(str)
df_clean["Description"] = df_clean["Description"].astype("string")
df_clean["Country"] = df_clean["Country"].astype("string")

df_clean["Quantity"] = pd.to_numeric(
    df_clean["Quantity"],
    errors="coerce"
)

df_clean["Price"] = pd.to_numeric(
    df_clean["Price"],
    errors="coerce"
)

df_clean["Customer ID"] = pd.to_numeric(
    df_clean["Customer ID"],
    errors="coerce"
)
```

```python
df_clean["InvoiceDate"] = pd.to_datetime(
    df_clean["InvoiceDate"],
    format="%d-%m-%Y %H:%M",
    errors="coerce"
)
```

```python
print(df_clean["InvoiceDate"].dtype)
print(df_clean["InvoiceDate"].isna().sum())
```

```python
df_clean["Year"] = df_clean["InvoiceDate"].dt.year
df_clean["Month"] = df_clean["InvoiceDate"].dt.month
df_clean["Day"] = df_clean["InvoiceDate"].dt.day
df_clean["Hour"] = df_clean["InvoiceDate"].dt.hour
df_clean["DayOfWeek"] = df_clean["InvoiceDate"].dt.day_name()
```

```python
df_clean["Revenue"] = (
    df_clean["Quantity"] * df_clean["Price"]
)
```

```python
df_clean["Revenue"].describe()
```

```python
def classify_transaction(row):

    invoice = str(row["Invoice"])
    stockcode = str(row["StockCode"]).upper()
    description = str(row["Description"]).upper()

    if stockcode == "AMAZONFEE":
        return "AMAZON_FEE"

    elif stockcode == "M":
        return "MANUAL_ADJUSTMENT"

    elif stockcode == "B":
        return "BAD_DEBT_ADJUSTMENT"

    elif invoice.startswith("C"):
        return "CANCELLATION"

    elif row["Quantity"] < 0:
        return "RETURN_OR_NEGATIVE_TRANSACTION"

    elif row["Price"] < 0:
        return "NEGATIVE_PRICE_ADJUSTMENT"

    elif row["Price"] == 0:
        return "ZERO_PRICE_TRANSACTION"

    else:
        return "SALE"
```

```python
df_clean["TransactionType"] = df_clean.apply(
    classify_transaction,
    axis=1
)
```
